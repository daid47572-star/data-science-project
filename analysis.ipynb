# Student Performance Analysis

## Research Question
**What are the key factors that influence student academic performance?**

This notebook analyzes the UCI Student Performance dataset to identify factors affecting student grades.

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
%matplotlib inline
```

```python
# Load dataset
df = pd.read_csv('data/student-mat.csv', sep=';')
print(f'Dataset shape: {df.shape}')
print(f'Number of students: {df.shape[0]}')
df.head()
```

```python
# Visualize grade distribution
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for idx, col in enumerate(['G1', 'G2', 'G3']):
    axes[idx].hist(df[col], bins=10, color=['skyblue', 'lightcoral', 'lightgreen'][idx], edgecolor='black')
    axes[idx].set_title(f'{col} Distribution')
    axes[idx].set_xlabel('Grade')
    axes[idx].set_ylabel('Frequency')
plt.tight_layout()
plt.savefig('visualizations/grade_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Correlation heatmap
numeric_cols = df.select_dtypes(include=[np.number]).columns
plt.figure(figsize=(14, 10))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', center=0, fmt='.2f', square=True)
plt.title('Correlation Matrix')
plt.tight_layout()
plt.savefig('visualizations/correlation_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Factor analysis
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

study_avg = df.groupby('studytime')['G3'].mean()
axes[0, 0].bar(study_avg.index, study_avg.values, color='steelblue')
axes[0, 0].set_title('Average Final Grade by Study Time')

axes[0, 1].scatter(df['absences'], df['G3'], alpha=0.5)
axes[0, 1].set_title('Final Grade vs Absences')

failure_avg = df.groupby('failures')['G3'].mean()
axes[1, 0].bar(failure_avg.index, failure_avg.values, color='lightcoral')
axes[1, 0].set_title('Average Final Grade by Past Failures')

axes[1, 1].scatter(df['G1'], df['G3'], alpha=0.5, color='green')
axes[1, 1].set_title('G1 vs G3 Correlation')

plt.tight_layout()
plt.savefig('visualizations/factor_analysis.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Key Findings

1. Study time has a positive correlation with final grades
2. Absences negatively impact performance
3. Past failures are strong predictors of future performance
4. G1 and G2 grades are strong predictors of G3
